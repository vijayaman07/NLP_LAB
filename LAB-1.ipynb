### 5. What is the difference between \d+  and  \w+ in regular expressions?

\d+ matches one or more digits (0–9). \d+ is used only for digits.

\w+ matches one or more non-word characters (Excluding letters, digits, and underscores (_)) . That is every special characters, spaces, etc..

```python
# CREATE A PYTHON LIST CONTAINING THE GIVEN TEXT CORPUS

corpus = [

    "<p>Data Science is AMAZING!!! visit https://datascience.com</p>",
    "Python 3.12 is released in 2025.",
    "NLP is used in AI.",
    "Contact us at info@gmail.com"
]
```

```python
#IMPORT THE REQUIRED LIBRARIES (re and BeautifulSoup)
import re
from bs4 import BeautifulSoup
```

```python
# CONVERT ALL TEXT TO LOWER CASE

for text in corpus:
    text = text.lower()
    print(text)
```

```python
lowercase_corpus = [text.lower() for text in corpus]
print(lowercase_corpus)
```

```python
# REMOVE HTML TAGS

html_removed = [
    BeautifulSoup(text, "html.parser").get_text()
    for text in lowercase_corpus
]
print(html_removed)
```

```python
# REMOVE URL'S

url_removed = [
    re.sub(r'https?://\S+',"",text)
    for text in html_removed
]
print(url_removed)
```

```python
# REMOVE EMAIL ADDRESSES

email_removed = [
    re.sub(r'\b[\w\.-]+@[\w\.-]+\.\w+\b','',text)
    for text in url_removed
]
print(email_removed)
```

```python
# REMOVE DIGITS.

digit_removed = [
    re.sub(r'\d+','', text)
    for text in email_removed
]
print(digit_removed)
```

```python
# REMOVE PUNCTUATION AND SPECIAL CHARACTER.

punct_removed = [
    re.sub(r'\W+',' ', text)
    for text in digit_removed
]
print(punct_removed)
```

```python
# REMOVE EXTRA WHITE SPACES

cleaned_corpus = [
    re.sub(r'\s+',' ',text).strip()
    for text in punct_removed
]
print(cleaned_corpus)
```

```python
# STORE THE CLEANED TEXT IN A NEW LIST NAMED cleaned_corpus

print("The cleaned text has been stored in the list 'cleaned_corpus'.")
print(cleaned_corpus)
```

```python
# DISPLAY THE ORIGINAL CORPUS AND THE CLEANED CORPUS

print("Original Corpus:\n")
for sentence in corpus:
    print(sentence)

print("\nCleaned Corpus:\n")
for sentence in cleaned_corpus:
    print(sentence)
```

```python
# COMPARE THE TEXT BEFORE AND AFTER CLEANING

print("Comparison Before and After Cleaning:\n")

for original,cleaned in zip(corpus,cleaned_corpus):
    print("Original :",original)
    print("Cleaned :", cleaned)
    print("-"*60)
```

```python
# TAKE OUR OWN DATA SET

reviews = [
    "My mans got eyes in the back of his head 👀 #restaurant #foodies #fypage #viralvideos",
    "Lots of Protein 🍔🥩 Contact denntt@meatshop.com",
    "Visit https://www.facebook.com/groups/1019064455361281/?ref=share&mibextid=NSMWBT",
    "fly high and rise ⭐  #studyreels #fypシ゚viralシ #studymotivation",
    "Wei Jiang W8606 Black Apple 🛒 Buy here 👉https://s.click.aliexpress.com/e/_c3HC0c4L 🤖 Transformers "
    
]

print("Original Reviews:\n")

for review in reviews:
    print(review)
```

```python
# IMPORT LIBRARIES

import re
from bs4 import BeautifulSoup
```

```python
# CLEAN THE DATA/ REVIEWS
cleaned_reviews =[]

for review in reviews:

    # convert to lower case
    review = review.lower()

    # Remove HTML Tags
    review =  BeautifulSoup(review,"html.parser").get_text()

    #Remove URL'S
    review = re.sub(r'https?://\S+|www\.\S+',' ',review)

    #Remove EMAIL ADDRESS
    review = re.sub(r'\b[\w\.-]+@[\w\.-]+\.\w+\b',' ', review)

    #Remove digits
    review = re.sub(r'\d+',' ', review)

    #Remove Punctuation and Special Characters
    review = re.sub(r'\W+',' ',review)

    #Remove Extra Spaces
    review = re.sub(r'\s+',' ',review).strip()

    cleaned_reviews.append(review)
```

```python
# COMPARE BEFORE AND AFTER CLEANING

print("Comparing of Original and Cleaned Reviews:\n")

for original,cleaned in zip(reviews,cleaned_reviews):

    print("Original Review:")
    print(original)

    print("\nCleaned Review:")
    print(cleaned)

    print("-"*80)
```
