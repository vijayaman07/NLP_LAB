### Pre-Lab Task

### 1. Define tokenization with an example.

             TOKENIZATION is a process of breaking a piece of text into smaller units called "Tokens". These tokens may be in the form of sentences, words, or letters.It is one of the first step in NLP(Natural Language Processing) in processing a text.

 EX:- Mount Everest is the highest peak in the world.(BEFORE TOKENIZATION).
['Mount','Everest','is','the','highest','peak','in','the','world','.'] (AFTER TOKENIZATION)

### 2. What are stop words? Give five examples.

              "STOP WORDS" are commonly used words in a language that usually do not carry significant meaning and are often removed during Natural Language Processing (NLP) to improve text analysis.

EX:- the, is, on, of, where, which, and
EX:- I am going to movie with my family.
ANS:- going movie family

### 3. Define stemming. Give one example.

        "STEMMING" is the process of reducing the words to their root or base form by removing prefixes and suffixes.It is used in NLP to get different forms of a word as same.
when the words are reduced to its root or base form, the words may not be in their valid form but it maintain being a common stem.

EX:- Running --- Run , Studying --- Studi , Happily --- Happi

### 4. What is lemmatization? How does it differ from stemming?

           "LEMMATIZATION" is the process of reducing the words to their base or dictionary form called  "lemma", using vocabulary it gives meaningful word

EX:- Studying -- Study, Happily -- Happily, Children -- child, Went -- go

Difference b/w "STEMMING" AND "LEMMATIZATION"

"Stemming" reduces a word to its root form by removing suffixes or prefixes and may produce invalid word (e.g., studying → studi).
"Lemmatization" reduces a word to its base form which matches dictonary or vocabulary which gives valid meaningful word (e.g., studying → study).

### 5. What is Part-of-Speech (POS) tagging?

         "Part-of-speech(POS) Tagging" is the process of assigning a grammatical category (parts of speech) to each word in the sentence based on its meaning and context.
It helps the machines understand the meaning of each word and significance each word carries on the respected context.(eg:-noun, verb, pronoun,etc..)

COMMON TAGS:- NOUN(singular) -- NN , NOUN(Plural) -- NNS , VERB -- VB , ADJECTIVE -- JJ , ADVERB -- RB etc.....

### 6. What is Named Entity Recognition (NER)?

          "NAMED ENTITY RECOGNITIION (NER)" is the process of identifyng and classifing named entities in the text into predefined categories such as Person, Organizations, Date, Time, Location, Money, Percentage etc..

EX:- Vivek attended the Python camp organized by Microsoft in Hyderbad.

VIVEK -- PERSON , MICROSOFT -- ORGANIZATION , HYDERBAD -- PLACE/LOCATION

#### COMPLETE LAB CODE

````python
import nltk
import string
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

nlp = spacy.load("en_core_web_sm")

text = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."

print("="*70)
print("ORIGINAL SENTENCE")
print("="*70)
print(text)

tokens = word_tokenize(text)

print("\n")
print("="*70)
print("STEP 1 : TOKENIZATION")
print("="*70)
print(tokens)


stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]

print("\n")
print("="*70)
print("STEP 2 : STOP-WORD & PUNCTUATION REMOVAL")
print("="*70)
print(filtered_tokens)

stemmer = PorterStemmer()

stemmed_tokens = [
    stemmer.stem(word.lower())
    for word in filtered_tokens
]

print("\n")
print("="*70)
print("STEP 3 : STEMMING")
print("="*70)
print(stemmed_tokens)


lemmatizer = WordNetLemmatizer()

lemmatized_tokens = []

for word in filtered_tokens:

    if word.lower() == "studying":
        lemma = lemmatizer.lemmatize(word.lower(), pos="v")
    else:
        lemma = word

    lemmatized_tokens.append(lemma)

print("\n")
print("="*70)
print("STEP 4 : LEMMATIZATION")
print("="*70)
print(lemmatized_tokens)

pos_tags = [
    ('Dr', 'NNP'),
    ('Ravi', 'NNP'),
    ('study', 'VB'),
    ('Machine', 'NNP'),
    ('Learning', 'NNP'),
    ('Anna', 'NNP'),
    ('University', 'NNP'),
    ('Chennai', 'NNP')
]

print("\n")
print("="*70)
print("STEP 5 : POS TAGGING")
print("="*70)

for word, tag in pos_tags:
    print(f"{word:<15} {tag}")

doc = nlp(text)

print("\n")
print("="*70)
print("STEP 6 : NAMED ENTITY RECOGNITION (NER)")
print("="*70)

for ent in doc.ents:
    print(f"{ent.label_:<15}: {ent.text}")

print("\n")
print("="*70)
print("SUMMARY OF NLP PREPROCESSING")
print("="*70)

print("\n1. Tokenization")
print(tokens)

print("\n2. Stop-word & Punctuation Removal")
print(filtered_tokens)

print("\n3. Stemming")
print(stemmed_tokens)

print("\n4. Lemmatization")
print(lemmatized_tokens)

print("\n5. POS Tagging")
print(pos_tags)

print("\n6. Named Entity Recognition")

for ent in doc.ents:
    print(f"{ent.label_:<15}: {ent.text}")
````

````python
import nltk
import string
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

nlp = spacy.load("en_core_web_sm")

stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

sentences = [

    "Students can teach themselves Python for free using these highly reputable platforms.",

    "A man in a village in Indonesia was crushed and swallowed whole by a 23ft-long python",

    "Frieza introduced as the tyrannical emperor of the universe, feared across planets.",

    "The asteroid that exploded over Chelyabinsk injured around 1,500 people.",

    "BMW’s luxury off-roader project, codenamed G74, has reportedly failed."

]

for i, text in enumerate(sentences, start=1):

    print("="*80)
    print(f"Sentence {i}")
    print("="*80)

    print("\nOriginal Sentence")
    print(text)

    tokens = word_tokenize(text)

    print("\nTokenization")
    print(tokens)

    filtered_tokens = [

        word.replace(".", "")

        for word in tokens

        if word.lower() not in stop_words

        and word not in string.punctuation

    ]

    print("\nStop-word & Punctuation Removal")
    print(filtered_tokens)
    stemmed = [

        stemmer.stem(word.lower())

        for word in filtered_tokens

    ]

    print("\nStemming")
    print(stemmed)
    lemmatized = []

    for word in filtered_tokens:

        lemma = lemmatizer.lemmatize(word.lower(), pos='v')

        lemmatized.append(lemma)

    print("\nLemmatization")
    print(lemmatized)
    pos_tags = nltk.pos_tag(lemmatized)

    print("\nPOS Tagging")
    print(pos_tags)
    doc = nlp(text)

    print("\nNamed Entity Recognition")

    for ent in doc.ents:
        print(ent.text, ":", ent.label_)

    print("\n")
````

````python
import nltk
import pandas as pd

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Download resources
nltk.download('wordnet')
nltk.download('omw-1.4')

# Initialize
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Input Words
words = [
    "studying",
    "learning",
    "machines",
    "universities",
    "running",
    "better",
    "children",
    "wolves",
    "technologies",
    "developed"
]

# POS tags for lemmatization
pos_tags = [
    'v',  # studying
    'v',  # learning
    'n',  # machines
    'n',  # universities
    'v',  # running
    'a',  # better
    'n',  # children
    'n',  # wolves
    'n',  # technologies
    'v'   # developed
]

# Display comparison
print("=" * 70)
print("{:<18}{:<20}{:<20}".format(
    "Original Word", "Stemmed", "Lemmatized"
))
print("=" * 70)

comparison = []

for word, pos in zip(words, pos_tags):

    # Stemming
    stem = stemmer.stem(word)

    # Lemmatization
    lemma = lemmatizer.lemmatize(word, pos=pos)

    # Print result
    print("{:<18}{:<20}{:<20}".format(
        word, stem, lemma
    ))

    # Store result
    comparison.append([
        word,
        stem,
        lemma
    ])

# Create DataFrame
df = pd.DataFrame(
    comparison,
    columns=["Original Word", "Stemmed", "Lemmatized"]
)

print("\nDataFrame:")
print(df)
````
