```python
import re
import numpy as np
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")

# Corpus
corpus = [
    "I love machine learning",
    "I enjoy artificial intelligence",
    "Cats drink milk",
    "Dogs drink milk"
]

print("\nInput Corpus:")
for i, sentence in enumerate(corpus, 1):
    print(f"S{i}: {sentence}")
```

```python
# Tokenization
tokenized_corpus = [
    re.findall(r'\b[a-zA-Z]+\b', sentence.lower())
    for sentence in corpus
]

print("Tokenized Corpus:")
for i, sentence in enumerate(tokenized_corpus, 1):
    print(f"S{i}: {sentence}")
```

```python
# Train Word2Vec model
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,
    workers=1,
    epochs=300,
    seed=42
)

print("Word2Vec model trained successfully!")
print("Vector Size:", model.vector_size)
```

```python
print("Word Vectors:")

for word in ["i", "love", "machine", "learning",
             "enjoy", "artificial", "intelligence"]:
    print(f"\n{word}:")
    print(model.wv[word])
```

```python
# Function to create sentence embedding
def sentence_embedding(sentence):
    words = re.findall(r'\b[a-zA-Z]+\b', sentence.lower())
    vectors = [model.wv[word] for word in words]
    return np.mean(vectors, axis=0)

# Create sentence embeddings
sentence1 = sentence_embedding("I love machine learning")
sentence2 = sentence_embedding("I enjoy artificial intelligence")

print("Sentence 1 Embedding:")
print(sentence1)

print("\nSentence 2 Embedding:")
print(sentence2)

print("\nEmbedding Dimension:", sentence1.shape)
```

```python
# Calculate cosine similarity
similarity = cosine_similarity(
    sentence1.reshape(1, -1),
    sentence2.reshape(1, -1)
)[0][0]

print("Cosine Similarity:", round(similarity, 4))
```

```python
print("\n" + "=" * 55)
print("       WORD2VEC SENTENCE SIMILARITY")
print("=" * 55)

print("Sentence 1 :", "I love machine learning")
print("Sentence 2 :", "I enjoy artificial intelligence")

print("-" * 55)

print("Vector Dimension :", sentence1.shape[0])
print("Cosine Similarity :", round(similarity, 2))

print("-" * 55)

if similarity >= 0.75:
    print("Result : Highly Similar")
elif similarity >= 0.50:
    print("Result : Moderately Similar")
elif similarity >= 0.25:
    print("Result : Slightly Similar")
else:
    print("Result : Low Similarity")

print("=" * 55)
```
